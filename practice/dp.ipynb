## 1. Fibonacci Sequence

### [Climbing Stairs](https://leetcode.com/problems/climbing-stairs/description/)

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?


**Example 1**:

Input: n = 2

Output: 2

Explanation: 

There are two ways to climb to the top.
- 1 step + 1 step
- 2 steps

**Example 2**:

Input: n = 3

Output: 3

Explanation:

There are three ways to climb to the top.
- 1 step + 1 step + 1 step
- 1 step + 2 steps
- 2 steps + 1 step


**Constraints**:
- 1 <= n <= 45

```python
def climbStairs_op(self, n: int) -> int:
    '''
        Time complexity: O(N)
        Space complexity: O(1)
    '''

    # Base case: If there are only 1 or 2 steps
    if n in {1, 2}:
        return n
    
    # Step counters
    prev, curr = 1, 2

    # Iterate through the remaining steps
    for _ in range(n - 2):
        # Update step counters using the Fibonacci sequence logic
        prev, curr = curr, curr + prev
    
    return curr

def climbStairs(self, n: int) -> int:
    '''
        Time complexity: O(N)
        Space complexity: O(N)
    '''

    # Base Case: If there are only 1 or 2 steps
    if n in {1, 2}:
        return n

    # Initialize a list to store the number of ways to reach each step
    res = [0] * n
    res[0] = 1
    res[1] = 2

    # Iterate through the remaining steps
    for i in range(2, n):
        # Update steps using the Fibonacci sequence logic
        res[i] = res[i - 1] + res[i - 2]

    return res[n - 1]
```

### [Fibonacci Number](leetcode.com/problems/fibonacci-number/description/)

The Fibonacci numbers, commonly denoted F(n) form a sequence, called the Fibonacci sequence, such that each number is the sum of the two preceding ones, starting from 0 and 1. That is,

    F(0) = 0, F(1) = 1
    F(n) = F(n - 1) + F(n - 2), for n > 1.

Given n, calculate F(n).


**Example 1**:

Input: n = 2
Output: 1
Explanation: F(2) = F(1) + F(0) = 1 + 0 = 1.

**Example 2**:

Input: n = 3
Output: 2
Explanation: F(3) = F(2) + F(1) = 1 + 1 = 2.

**Example 3**:

Input: n = 4
Output: 3
Explanation: F(4) = F(3) + F(2) = 2 + 1 = 3.


**Constraints**:
- 0 <= n <= 30

```python
def fib(self, n: int) -> int:
    '''
        Time complexity: O(N)
        Space complexity: O(1)
    '''
    # Base Case: n is either 0 or 1
    if n in {0, 1}:
        return n
    
    # Counters to track the previous and current Fibonacci numbers.
    prev, curr = 0, 1
    for _ in range(n - 1):
        # Update the counters.
        prev, curr = curr, prev + curr

    return curr
```

### [Min Cost Climbing Stairs](https://leetcode.com/problems/min-cost-climbing-stairs/description/)
You are given an integer array cost where cost[i] is the cost of ith step on a staircase. Once you pay the cost, you can either climb one or two steps.

You can either start from the step with index 0, or the step with index 1.

Return the minimum cost to reach the top of the floor.


**Example 1**:

Input: cost = [10,15,20]
Output: 15
Explanation: 
- You will start at index 1.
- Pay 15 and climb two steps to reach the top.
- The total cost is 15.

**Example 2**:

Input: cost = [1,100,1,1,1,100,1,1,100,1]
Output: 6
Explanation: 
- You will start at index 0.
- Pay 1 and climb two steps to reach index 2.
- Pay 1 and climb two steps to reach index 4.
- Pay 1 and climb two steps to reach index 6.
- Pay 1 and climb one step to reach index 7.
- Pay 1 and climb two steps to reach index 9.
- Pay 1 and climb one step to reach the top.
- The total cost is 6.

Constraints:

- 2 <= cost.length <= 1000
- 0 <= cost[i] <= 999

```python
def minCostClimbingStairs(self, cost: list[int]) -> int:
    '''
        Time complexity: O(n)
        Space complexity: O(1)
    '''

    # Initialize variables and step counters
    n = len(cost)
    prev, curr = cost[0], cost[1]

    # Base case: If there are only two steps
    if n <= 2:
        return min(prev, curr)
    
    # Iterate through the remaining steps
    for i in range(2, n):
        # Update the counters
        prev, curr = curr, cost[i] + min(curr, prev)
    
    return min(prev, curr)
```

## 2. Kadane's Algorithm

### [Maximum Subarray](https://leetcode.com/problems/maximum-subarray/description/)

Given an integer array nums, find the *subarray* with the largest sum, and return its sum.

**Example 1:**

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]

Output: 6

Explanation: The subarray [4,-1,2,1] has the largest sum 6.

**Example 2:**

Input: nums = [1]

Output: 1

Explanation: The subarray [1] has the largest sum 1.

**Example 3:**

Input: nums = [5,4,-1,7,8]

Output: 23

Explanation: The subarray [5,4,-1,7,8] has the largest sum 23.


**Constraints:**

- 1 <= nums.length <= $10^5$
- $-10^4$ <= nums[i] <= $10^4$

**Follow up:** If you have figured out the O(n) solution, try coding another solution using the divide and conquer approach, which is more subtle.

```python
def maxSubArray(self, nums: list[int]) -> int:
    '''
        Time Complexity: O(n)
        Space Complexity: O(1)
    '''
    # Initialize counters
    localMax = 0
    globalMax = nums[0]

    for num in nums:
        # Update the local and global maximums
        localMax = max(localMax + num, num)
        globalMax = max(globalMax, localMax)

    return globalMax
```

### [Maximum Sum Circular Subarray](https://leetcode.com/problems/maximum-sum-circular-subarray/description/)

Given a circular integer array nums of length n, return the maximum possible sum of a non-empty subarray of nums.

A circular array means the end of the array connects to the beginning of the array. Formally, the next element of `nums[i]` is `nums[(i + 1) % n]` and the previous element of `nums[i]` is `nums[(i - 1 + n) % n]`.

A subarray may only include each element of the fixed buffer nums at most once. Formally, for a subarray `nums[i], nums[i + 1], ..., nums[j]`, there does not exist `i <= k1, k2 <= j with k1 % n == k2 % n`.
 

**Example 1:**

Input: nums = [1,-2,3,-2]

Output: 3

Explanation: Subarray [3] has maximum sum 3.


**Example 2:**

Input: nums = [5,-3,5]

Output: 10

Explanation: Subarray [5,5] has maximum sum 5 + 5 = 10.


**Example 3:**

Input: nums = [-3,-2,-3]

Output: -2

Explanation: Subarray [-2] has maximum sum -2.


**Constraints:**

- n == nums.length
- 1 <= n <= $3 * 10^4$
- $-3 * 10^4$ <= nums[i] <= $3 * 10^4$

```python
def maxSubarraySumCircular(self, nums: list[int]) -> int:
    '''
        Time Complexity: O(n)
        Space Complexity: O(1)
    '''
    # Initialise the tracking counters
    currMax = currMin = total = 0
    maxSum = minSum = nums[0]

    for num in nums:
        # Update the max-track counters
        currMax = max(currMax + num, num)
        maxSum = max(maxSum, currMax)

        # Update the min-track counters
        currMin = min(currMin + num, num)
        minSum = min(minSum, currMin)

        total += num
    
    # Return the maximum of the two sums
    return max(maxSum, total - minSum) if maxSum > 0 else maxSum
```

### [Maximum Product Subarray](https://leetcode.com/problems/maximum-product-subarray/description/)

Given an integer array nums, find a *subarray* that has the largest product, and return the product.

The test cases are generated so that the answer will fit in a 32-bit integer.


**Example 1:**

Input: nums = [2,3,-2,4]

Output: 6

Explanation: [2,3] has the largest product 6.


**Example 2:**

Input: nums = [-2,0,-1]

Output: 0

Explanation: The result cannot be 2, because [-2,-1] is not a subarray. 


**Constraints:**

- 1 <= nums.length <= $2 * 10^4$
- -10 <= nums[i] <= 10
- The product of any subarray of nums is guaranteed to fit in a 32-bit integer.

```python
def maxProduct(self, nums: List[int]) -> int:
    '''
        Time complexity: O(N)
        Space complexity: O(1)
    '''

    # Initialize tracking counters
    n = len(nums)
    res = nums[0]
    pre = suff = 1

    for i in range(n):
        # Update tracking counters
        pre *= nums[i]
        suff *= nums[n - 1 - i]

        res = max(res, max(pre, suff))

        # Reset tracking counters to 1 if they become zero
        pre = 1 if not pre else pre
        suff = 1 if not suff else suff

    return res
```

## 3. 0/1 Knapsack

### [Partition Equal Subset Sum](https://leetcode.com/problems/partition-equal-subset-sum/description/)

Given an integer array nums, return true if you can partition the array into two subsets such that the sum of the elements in both subsets is equal or false otherwise.


**Example 1:**

Input: nums = [1,5,11,5]

Output: true

Explanation: The array can be partitioned as [1, 5, 5] and [11].


**Example 2:**

Input: nums = [1,2,3,5]

Output: false

Explanation: The array cannot be partitioned into equal sum subsets.
 

**Constraints:**

- 1 <= nums.length <= 200
- 1 <= nums[i] <= 100

```python
def canPartition(self, nums: list[int]) -> bool:
    '''
        Time Complexity: O(N * TargetSum)
        Space Complexity: O(TargetSum)
    '''

    # Compute the sum of the array and check if it is even
    arrSum = sum(nums)
    if arrSum % 2:
        return False
    
    # Compute the target sum that each subset should have
    targetSum = arrSum // 2

    # Initialize the DP array with False values and set the base case to True
    dp = [False] * (targetSum + 1)
    dp[0] = True

    for num in nums:
        for currSum in range(num, targetSum + 1)[::-1]:
            # Update the DP array based on whether we can form a subset for the current number or not
            dp[currSum] = dp[currSum] or dp[currSum - num]

    return dp[targetSum]
```

### [Target Sum](leetcode.com/problems/target-sum/description/)

You are given an integer array nums and an integer target.

You want to build an expression out of nums by adding one of the symbols '+' and '-' before each integer in nums and then concatenate all the integers.

For example, if nums = [2, 1], you can add a '+' before 2 and a '-' before 1 and concatenate them to build the expression "+2-1".

Return the number of different expressions that you can build, which evaluates to target.

 

**Example 1:**

Input: nums = [1,1,1,1,1], target = 3

Output: 5

Explanation: There are 5 ways to assign symbols to make the sum of nums be target 3.
$$
-1 + 1 + 1 + 1 + 1 = 3\\

+1 - 1 + 1 + 1 + 1 = 3\\

+1 + 1 - 1 + 1 + 1 = 3\\

+1 + 1 + 1 - 1 + 1 = 3\\

+1 + 1 + 1 + 1 - 1 = 3\\
$$

**Example 2:**

Input: nums = [1], target = 1

Output: 1


**Constraints:**

- 1 <= nums.length <= 20
- 0 <= nums[i] <= 1000
- 0 <= sum(nums[i]) <= 1000
- -1000 <= target <= 1000

**The Problem Transformation: From +/- to Subsets**

First, let's re-state the two problems in plain English:

- Last Problem (Partition): Find if you can split nums into two piles with an equal sum.
- This Problem (Target Sum): Find how many ways you can put a + or a - in front of each number in nums so that they add up to target.

Putting a + or - in front of each number is the same as splitting all the numbers into two groups: a "positive" group (P) that gets added, and a "negative" group (N) that gets subtracted.

So, the goal of this new problem is to find:
$sum(P) - sum(N) = target$

This doesn't look like our last problem yet. But we know something else that's always true: if we add the sums of the two groups, we get the total sum of the original list.

$sum(P) + sum(N) = totalSum (where totalSum = sum(nums))$

Now we have two simple equations. Let's do some middle-school algebra!
- $sum(P) - sum(N) = target$
- $sum(P) + sum(N) = totalSum$

If we add these two equations together, the - sum(N) and + sum(N) cancel each other out:

$$
(sum(P) - sum(N)) + (sum(P) + sum(N)) = target + totalSum\\
2 * sum(P) = target + totalSum\\
$$

And now, we solve for sum(P):
$sum(P) = (totalSum + target) / 2$

**What This Magic Formula Means**

This final equation is the key! It tells us that:

The task of finding ways to assign + and - to get a target is exactly the same as finding a subset P whose elements sum up to a new goal: (totalSum + target) / 2.

Once you find such a subset P, all the other numbers in nums automatically form the N group, and the math will work out perfectly to give you the target.

```python
def findTargetSumWays(self, nums: list[int], target: int) -> int:
    '''
        Time Complexity: O(N * subTarget)
        Space Complexity: O(subTarget)
    '''

    # Compute the subTarget amount and check if it's valid
    subTarget = (sum(nums) + target)

    if subTarget % 2 or abs(target) > sum(nums):
        return 0

    # Initialise the counters
    subTarget //= 2
    dp = [0] * (subTarget + 1)
    dp[0] = 1

    for num in nums:
        for currSum in range(num, subTarget + 1)[::-1]:
            # Update the counter for the current sum
            dp[currSum] += dp[currSum - num]

    return dp[subTarget]
```

### [Last Stone Weight II](https://leetcode.com/problems/last-stone-weight-ii/description/)

You are given an array of integers stones where stones[i] is the weight of the ith stone.

We are playing a game with the stones. On each turn, we choose any two stones and smash them together. Suppose the stones have weights x and y with x <= y. The result of this smash is:

- If x == y, both stones are destroyed, and
- If x != y, the stone of weight x is destroyed, and the stone of weight y has new weight y - x.

At the end of the game, there is at most one stone left.

Return the smallest possible weight of the left stone. If there are no stones left, return 0.


**Example 1:**

Input: stones = [2,7,4,1,8,1]

Output: 1

Explanation:
- We can combine 2 and 4 to get 2, so the array converts to [2,7,1,8,1] then,
- we can combine 7 and 8 to get 1, so the array converts to [2,1,1,1] then,
- we can combine 2 and 1 to get 1, so the array converts to [1,1,1] then,
- we can combine 1 and 1 to get 0, so the array converts to [1], then that's the optimal value.


**Example 2:**

Input: stones = [31,26,33,21,40]

Output: 5
 

**Constraints:**

- 1 <= stones.length <= 30
- 1 <= stones[i] <= 100

**Connecting to Previous Problems**

Now, let's see how this relates to the problems you've already solved:

- **Partition Equal Subset Sum**: In this problem, we asked, "Can we make the difference `abs(sum(S1) - sum(S2))` exactly zero?" It was a True or False question.

- **Target Sum**: Here, we asked, "How many ways can we make the difference `sum(S1) - sum(S2)` exactly equal to a specific target?" This was a counting problem.

- **Last Stone Weight II**: Now, we are asking, "What is the minimum possible value of `abs(sum(S1) - sum(S2))`?" This is an optimization problem.

All three problems revolve around partitioning a set into two subsets. This new problem is about finding the best partition to make the two subsets as close in sum as possible.


**The Dynamic Programming Solution**

The solution uses the same logic we've established. Let totalSum be the sum of all stones. We are trying to find a subset of stones whose sum, let's call it subsetSum, is as close as possible to      `totalSum / 2`.

If we can find such a subset S1 with sum subsetSum, then the other subset S2 will have a sum of totalSum - subsetSum. The difference between them will be:

$abs((totalSum - subsetSum) - subsetSum) = abs(totalSum - 2 * subsetSum)$

To minimize this difference, we need to make subsetSum as large as possible, without exceeding totalSum / 2.

```python
def lastStoneWeightII(self, stones: list[int]) -> int:
    '''
        Time Complexity: O(N * Target)
        Space Complexity: O(Target)
    '''

    # Initialize DP and compute target value (for minimization)
    target = sum(stones) // 2
    dp = [False] * (target + 1)
    dp[0] = True

    for stone in stones:
        for currSum in range(stone, target + 1)[::-1]:
            # Update DP table to reflect current stone inclusion status.
            dp[currSum] |= dp[currSum - stone]

    for currSum in range(target + 1)[::-1]:
        # Return the minimum difference between totalSum and twice subset
        if dp[currSum]:
            return sum(stones) - 2 * currSum
```
